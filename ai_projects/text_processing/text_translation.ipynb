```python
import tensorflow as tf
import transformers
from transformers import pipeline
```

```python
# English to German 
# When using Opus we will need to use specific directional translations
opus_model = "Helsinki-NLP/opus-mt-en-de"

translate_pipeline = pipeline(task="translation", model=opus_model, framework="tf", torch_dtype=None, device=0)
```

```python
input_text = ['Hello, How are you ?', ' Can i offer a cup of tea ?']

output_text = translate_pipeline(input_text)

print('First Input Text:', input_text[0])
print('First Output Text:', output_text[0])

print('Second Input Text:', input_text[1])
print('Second Output Text:', output_text[1])
```
